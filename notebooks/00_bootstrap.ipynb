# 00 — Per-session bootstrap

Run this once at the start of every Colab session. It:

1. Mounts Google Drive (data + checkpoints live there).
2. Clones (or pulls) the GitHub repo into `/content/bangla-vlm-lora`.
3. Installs Python dependencies.
4. Verifies `paths.py` resolves to Drive.

**Prerequisites (one-time):** Drive folder `MyDrive/bangla-vlm-lora/` exists; `gh auth login` done in this session if you intend to push.

## 1. Mount Drive

```python
from google.colab import drive
drive.mount('/content/drive')
```

## 2. Clone or pull the repo

```python
import os, subprocess
REPO_URL = 'https://github.com/BlackBeard009/bangla-vlm-lora.git'
REPO_DIR = '/content/bangla-vlm-lora'

if os.path.isdir(REPO_DIR + '/.git'):
    print('Repo already present — pulling latest.')
    subprocess.run(['git', '-C', REPO_DIR, 'pull', '--ff-only'], check=True)
else:
    subprocess.run(['git', 'clone', REPO_URL, REPO_DIR], check=True)

%cd $REPO_DIR
!git log --oneline -1
```
